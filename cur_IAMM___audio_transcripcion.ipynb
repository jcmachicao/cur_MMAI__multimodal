```python
import os
from pydub import AudioSegment

# Ensure audios_dir is defined (it should be from a previous cell)
# If not, uncomment and define it:
# audios_dir = 'spanish_audio_samples'
# if not os.path.exists(audios_dir):
#     os.makedirs(audios_dir)

processed_uploaded_files = []

print(f"Processing uploaded audio files and moving to '{audios_dir}'...")

for filename in uploaded.keys():
    source_path = os.path.join('/content/', filename) # Uploaded files are typically in /content/

    # Define supported extensions for conversion
    supported_extensions = ('.mp3', '.ogg', '.wav', '.m4a')

    if filename.lower().endswith(supported_extensions):
        base_name = os.path.splitext(filename)[0]
        target_path = os.path.join(audios_dir, f'{base_name}.wav')

        try:
            # Load audio using pydub
            audio = AudioSegment.from_file(source_path)
            # Export to WAV format
            audio.export(target_path, format='wav')
            processed_uploaded_files.append(target_path);
            print(f"Processed and converted '{filename}' to '{target_path}'")
        except Exception as e:
            print(f"Error processing '{filename}': {e}")
    else:
        print(f"Skipping '{filename}': Unsupported file format. Please upload .mp3, .ogg, .wav, or .m4a files.")


if processed_uploaded_files:
    print(f"\nSuccessfully processed {len(processed_uploaded_files)} audio file(s) and moved to '{audios_dir}'.")
    print("Processed files:")
    for f in processed_uploaded_files:
        print(f"- {f}")
else:
    print("\nNo supported audio files were processed. Please upload appropriate files.")

# List all files in the target directory to confirm
print(f"\nContents of '{audios_dir}':")
for f in os.listdir(audios_dir):
    print(f"- {f}")
```

**Reasoning**:
Now that at least one audio file has been successfully uploaded and converted, the next step is to create a degraded version of it, as per the subtask instructions. I will use `pydub` to add noise to the existing audio file.



```python
import os
from pydub import AudioSegment
from pydub.generators import WhiteNoise

# Get the list of processed audio files. Assuming there's at least one.
# If processed_uploaded_files is empty, this step will need to be re-evaluated.
if processed_uploaded_files:
    base_audio_path = processed_uploaded_files[0]
    base_audio_filename = os.path.basename(base_audio_path)
    print(f"Using '{base_audio_filename}' as the base for degradation.")

    # Load the base audio
    base_audio = AudioSegment.from_file(base_audio_path)

    # --- Degradation 1: Add white noise ---
    # Generate 1-minute white noise (or duration of base audio if shorter)
    noise = WhiteNoise().to_audio_segment(duration=len(base_audio))

    # Overlay noise on the base audio. Adjust dBFS for desired noise level.
    # A negative value means the noise will be quieter relative to the base audio.
    # Example: -10 dB is quite noticeable, -20 dB is more subtle.
    degraded_audio_noise = base_audio.overlay(noise - 20) # -20 dBFS noise

    # Define path for the degraded audio file
    degraded_filename_noise = os.path.splitext(base_audio_filename)[0] + '_degraded_noise.wav'
    degraded_audio_path_noise = os.path.join(audios_dir, degraded_filename_noise)

    # Export the degraded audio
    degraded_audio_noise.export(degraded_audio_path_noise, format='wav')
    print(f"Created degraded version (with noise): '{degraded_audio_path_noise}'")

    # --- Degradation 2: Reduce quality (lower bitrate/sample rate) ---
    # This often involves re-exporting with different parameters.
    degraded_filename_quality = os.path.splitext(base_audio_filename)[0] + '_degraded_quality.wav'
    degraded_audio_path_quality = os.path.join(audios_dir, degraded_filename_quality)

    # Example: Export at a lower sample rate (e.g., 8kHz) and lower bitrate if applicable
    # Note: pydub doesn't directly control 'bitrate' for WAV as it's uncompressed.
    # We can simulate by changing frame rate and converting to mono.
    degraded_audio_quality = base_audio.set_frame_rate(8000).set_channels(1)
    degraded_audio_quality.export(degraded_audio_path_quality, format='wav')
    print(f"Created degraded version (lower quality): '{degraded_audio_path_quality}'")

    print(f"\nAll audio files in '{audios_dir}':")
    for f in os.listdir(audios_dir):
        print(f"- {f}")
else:
    print("No base audio file found to degrade. Please ensure at least one audio file was uploaded and processed.")
```

```python
import os
from pydub import AudioSegment
from pydub.generators import WhiteNoise

# Get the list of all original audio files (excluding previously degraded ones)
# We'll use processed_uploaded_files as it contains all original WAVs in audios_dir
original_base_audios = [f for f in processed_uploaded_files if not ('_degraded_' in f)]

print("Creating degraded versions for additional original audio files...")

for base_audio_path in original_base_audios:
    base_audio_filename = os.path.basename(base_audio_path)

    # Check if degraded versions already exist for this file to avoid re-processing
    degraded_noise_check_path = os.path.join(audios_dir, os.path.splitext(base_audio_filename)[0] + '_degraded_noise.wav')
    degraded_quality_check_path = os.path.join(audios_dir, os.path.splitext(base_audio_filename)[0] + '_degraded_quality.wav')

    if os.path.exists(degraded_noise_check_path) and os.path.exists(degraded_quality_check_path):
        print(f"Skipping '{base_audio_filename}': Degraded versions already exist.")
        continue

    print(f"Processing '{base_audio_filename}' for degradation.")

    try:
        # Load the base audio
        base_audio = AudioSegment.from_file(base_audio_path)

        # --- Degradation 1: Add white noise ---
        noise = WhiteNoise().to_audio_segment(duration=len(base_audio))
        degraded_audio_noise = base_audio.overlay(noise - 20) # -20 dBFS noise

        degraded_filename_noise = os.path.splitext(base_audio_filename)[0] + '_degraded_noise.wav'
        degraded_audio_path_noise = os.path.join(audios_dir, degraded_filename_noise)
        degraded_audio_noise.export(degraded_audio_path_noise, format='wav')
        print(f"  Created degraded version (with noise): '{degraded_audio_path_noise}'")

        # --- Degradation 2: Reduce quality (lower sample rate/mono) ---
        degraded_filename_quality = os.path.splitext(base_audio_filename)[0] + '_degraded_quality.wav'
        degraded_audio_path_quality = os.path.join(audios_dir, degraded_filename_quality)

        degraded_audio_quality = base_audio.set_frame_rate(8000).set_channels(1)
        degraded_audio_quality.export(degraded_audio_path_quality, format='wav')
        print(f"  Created degraded version (lower quality): '{degraded_audio_path_quality}'")

    except Exception as e:
        print(f"Error degrading '{base_audio_filename}': {e}")

print(f"\nAll audio files in '{audios_dir}' after final degradation processing:")
for f in os.listdir(audios_dir):
    print(f"- {f}")
```

```python
import whisper

print("Loading Whisper model... This may take a few minutes.")
# For Spanish, 'medium' or 'large' are recommended for better accuracy.
# 'small' can be used for faster processing if needed.
model = whisper.load_model("medium")
print("Whisper 'medium' model loaded successfully.")
```

**Reasoning**:
I need to transcribe each audio sample using the loaded Whisper model, measure the processing time, and store the results as per the subtask instructions.



```python
import os
import time

# Ensure audios_dir and model are available from previous steps

print(f"Starting transcription of audio files in '{audios_dir}'...")

audio_files_to_transcribe = []
for file in os.listdir(audios_dir):
    if file.endswith('.wav'):
        audio_files_to_transcribe.append(os.path.join(audios_dir, file))

if not audio_files_to_transcribe:
    print(f"No .wav audio files found in '{audios_dir}'. Please ensure files are present.")
else:
    print(f"Found {len(audio_files_to_transcribe)} audio files to transcribe.")

transcription_results = []

for i, audio_path in enumerate(audio_files_to_transcribe):
    filename = os.path.basename(audio_path)
    print(f"\n[{i+1}/{len(audio_files_to_transcribe)}] Transcribing: {filename}")

    start_time = time.time()
    try:
        result = model.transcribe(audio_path, language='es')
        transcription = result["text"]
        time_taken = time.time() - start_time

        print(f"  Transcription (first 100 chars): {transcription[:100]}...")
        print(f"  Time taken: {time_taken:.2f} seconds")

        transcription_results.append({
            "audio_path": audio_path,
            "filename": filename,
            "transcription": transcription,
            "time_taken_seconds": time_taken
        })
    except Exception as e:
        print(f"  Error transcribing {filename}: {e}")
        transcription_results.append({
            "audio_path": audio_path,
            "filename": filename,
            "transcription": "ERROR: " + str(e),
            "time_taken_seconds": -1 # Indicate error
        })

print("\nTranscription process completed.")
print(f"Total transcriptions recorded: {len(transcription_results)}")
```

## Ideas para Evaluar las Transcripciones

Evaluación cualitativa manual:

Escucha comparativa: Escucha cada audio original y compara la transcripción generada. Presta atención a la fidelidad del texto, la puntuación y la correcta identificación de palabras, especialmente en nombres propios o terminología específica.
Identificación de errores específicos: Busca tipos de errores comunes como omisiones, inserciones, sustituciones (por ejemplo, palabras similares pero incorrectas) y errores de homófonos. Anota si los errores son más frecuentes en audios degradados o con acentos específicos.
Contexto y coherencia: Evalúa si la transcripción mantiene el sentido general del discurso, incluso si hay pequeños errores. A veces, un error puntual no afecta la comprensión global.
Métricas de evaluación cuantitativas (si tienes transcripciones de referencia):

Word Error Rate (WER): Esta es la métrica más común para la evaluación de ASR. Mide el número de sustituciones, inserciones y eliminaciones necesarias para transformar la transcripción generada en la transcripción de referencia (ground truth). Un WER más bajo indica mejor rendimiento.
Character Error Rate (CER): Similar al WER, pero a nivel de caracteres, útil para idiomas con morfología compleja o cuando la granularidad a nivel de carácter es importante.
Precisión, Recall y F1-Score: Si separamos la evaluación en segmentos o palabras clave, estas métricas pueden ser útiles para ver la capacidad del modelo para identificar correctamente ciertos elementos.
Análisis de rendimiento según las características del audio:

Impacto de la degradación: Compara el WER (o evaluación cualitativa) de los audios originales con sus versiones degradadas (ruido, baja calidad). ¿Qué tipo de degradación afecta más el rendimiento de Whisper?
Impacto del acento: Compara el rendimiento entre los diferentes acentos. ¿Hay algún acento particular con el que Whisper tenga más dificultades o, por el contrario, mejor desempeño?
