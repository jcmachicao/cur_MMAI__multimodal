```python
from google.colab import userdata
api_user = userdata.get('OAIK_JCMV')
```

```python
client = OpenAI(api_key=api_user)
```

```python
url_building = "https://gestion.pe/resizer/v2/IFJRNUDHJRGOLABTNMBINOAN6E.jpg?auth=b4f8f5f3cc468145580f4cf7d71649b522de95b7f2c831d8f52dd4996b4236d3&width=1200&height=851&quality=75&smart=true"
```

```python
from openai import OpenAI
response = client.chat.completions.create(
    model="gpt-4o",  # o "gpt-4o-mini"
    messages=[
        {
            "role": "user",
            "content": [
                {"type": "text", "text": "¿Qué hay en esta imagen y qué colores destacan?"},
                {
                    "type": "image_url",
                    "image_url": {
                        "url": url_building
                    },
                },
            ],
        }
    ],
    max_tokens=300,
)

print(response.choices[0].message.content)
```
